```python
# First: check what files actually exist in the repo
from huggingface_hub import list_repo_files

files = list(list_repo_files("castorini/wura", repo_type="dataset"))
# Print files for hau to see the real paths
for f in files:
    if "hau" in f:
        print(f)
```

```python
import pandas as pd
from datasets import Dataset, concatenate_datasets
from huggingface_hub import hf_hub_download

LANGS = ["hau", "ibo", "yor"]
datasets_list = []

for lang in LANGS:
    path = hf_hub_download(
        repo_id="castorini/wura",
        filename=f"documents-v1.0/train/{lang}.jsonl",
        repo_type="dataset"
    )
    df = pd.read_json(path, lines=True)
    df["language"] = lang
    datasets_list.append(Dataset.from_pandas(df))
    print(f"✓ {lang}: {len(df):,} documents")

wura_nigerian = concatenate_datasets(datasets_list)
print(wura_nigerian)
print(wura_nigerian[0])
```
